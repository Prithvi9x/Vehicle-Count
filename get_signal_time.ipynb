```python
import json
import os
```

```python
weights = {
    "person": 0.5,
    "bicycle": 0.7,
    "motorcycle": 1,
    "car": 2,
    "bus": 3,
    "truck": 3.5,
    "ambulance": 10,
    "fire_truck": 10
}
```

```python
# Base parameters
base_time = 10        # Minimum green light time (in seconds)
scaling_factor = 2    
```

```python
def calculate_signal_timings(json_path):
    if not os.path.exists(json_path):
        print(f"File not found: {json_path}")
        return None

    with open(json_path, "r") as f:
        data = json.load(f)

    lane_data = data.get("lane_counts", {})
    signal_times = {}

    print(f"\nVehicle Data for Image: {data.get('image', 'unknown')}")
    print("-" * 60)

    for lane, vehicles in lane_data.items():
        print(f"\n{lane}:")

        total_weight = 0
        for v, c in vehicles.items():
            w = weights.get(v, 1)
            subtotal = w * c
            total_weight += subtotal
            print(f"  {v:<12} → count: {c:<2} × weight: {w:<4} = {subtotal:<5.2f}")

        signal_time = base_time + scaling_factor * total_weight

        # Emergency vehicle priority
        if any(v in vehicles for v in ["ambulance", "fire_truck"]):
            print("  Emergency vehicle detected → +10 sec priority")
            signal_time += 10

        print(f"  + Base time: {base_time} sec")
        print(f"  + Weighted contribution: {scaling_factor} × {total_weight:.2f} = {scaling_factor * total_weight:.2f}")
        print(f"  = Final signal time for {lane}: {signal_time:.2f} sec")

        signal_times[lane] = round(signal_time, 2)

    print("\n" + "=" * 60)
    print("FINAL SIGNAL TIMINGS (in seconds)")
    for lane, t in signal_times.items():
        print(f"  {lane}: {t} sec")
    print("=" * 60 + "\n")

    return signal_times
```

```python
json_path = "FirstFrame_annotate-1/output_images/Nd_O2_1654063379.3419924/output_frame22.json"
calculate_signal_timings(json_path)
```
